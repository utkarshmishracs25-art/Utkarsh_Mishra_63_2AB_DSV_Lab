```python
import pandas as pd
import os
```

```python
df = pd.read_csv("/content/StudentsPerformance.csv")
```

```python
print("Shape:", df.shape)
print()
print(df.columns.tolist())
print()
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.to_excel("StudentsPerformance.xlsx", index=False)
df.to_json("StudentsPerformance.json", orient="records", indent=4)
df.to_parquet("StudentsPerformance.parquet", index=False)

csv_df = pd.read_csv("StudentsPerformance.csv")
excel_df = pd.read_excel("StudentsPerformance.xlsx")
json_df = pd.read_json("StudentsPerformance.json")
parquet_df = pd.read_parquet("StudentsPerformance.parquet")

print("CSV Shape:", csv_df.shape)
print("Excel Shape:", excel_df.shape)
print("JSON Shape:", json_df.shape)
print("Parquet Shape:", parquet_df.shape)
```

```python
files = [
    "StudentsPerformance.csv",
    "StudentsPerformance.xlsx",
    "StudentsPerformance.json",
    "StudentsPerformance.parquet"
]
for file in files:
    size_kb = os.path.getsize(file)
    print(f"{file}: {size_kb:.2f} KB")
```
